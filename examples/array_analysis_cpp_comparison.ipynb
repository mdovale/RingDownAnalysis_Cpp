# Array analysis: C++ vs Python parity

From the repository root, generate references and compare:

```bash
python examples/python/export_array_reference.py
build/dev/examples/array_analysis_example --input-csv results/examples/array_analysis_py/shared_input.csv
python examples/python/compare_array_analysis.py \
  --py-json results/examples/array_analysis_py/from_t_and_data.json \
  --cpp-json results/examples/array_analysis_cpp/from_t_and_data.json \
  --waveform-rtol 0 --waveform-atol 0
```

Use `--plot results/examples/array_analysis_cpp/array_overlay.png` for a quick overlay figure.


````python
import subprocess
import sys
from pathlib import Path

ROOT = Path.cwd()
subprocess.check_call(
    [sys.executable, str(ROOT / "examples/python/export_array_reference.py")],
    cwd=ROOT,
)
subprocess.check_call(
    [
        str(ROOT / "build/dev/examples/array_analysis_example"),
        "--input-csv",
        str(ROOT / "results/examples/array_analysis_py/shared_input.csv"),
        "--output-dir",
        str(ROOT / "results/examples/array_analysis_cpp"),
    ],
    cwd=ROOT,
)
subprocess.check_call(
    [
        sys.executable,
        str(ROOT / "examples/python/compare_array_analysis.py"),
        "--py-json",
        str(ROOT / "results/examples/array_analysis_py/from_t_and_data.json"),
        "--cpp-json",
        str(ROOT / "results/examples/array_analysis_cpp/from_t_and_data.json"),
        "--waveform-rtol",
        "0",
        "--waveform-atol",
        "0",
        "--plot",
        str(ROOT / "results/examples/array_analysis_cpp/array_overlay.png"),
    ],
    cwd=ROOT,
)
print("Done.")
````
